**Regular Expression**

A regular expression is a pattern that the regular expression engine attempts to match in input text.

Python has a built-in package called 're', which can be used to work with Regular Expressions

findall	: Returns a list containing all matches

search : Returns a Match object if there is a match anywhere in the string

split : Returns a list where the string has been split at each match

sub : Replaces one or many matches with a string

```python
import re

#Return a list containing every occurrence of "ll":

txt = "Hi all How are you doing, Doing well"
x = re.findall("ll", txt) # return a list of occurrences
print(x)
print(len(x))

txt = "Hi all How are you doing, Doing well"
x = re.findall("chan", txt)
print(x)
```

```python
x = re.search("\s", txt) # \s : means space

print("The first white-space character is located in position:", x.start()) # it will give the position
print(x)

x = re.search("chan", txt)
print(x)
```

```python
txt = "Hi all How are you doing, Doing well"
x = re.split("ll", txt)
print(x)
```

```python
x = re.sub("\s", "\n", txt) # \n is new line
print(x)
```

```python
# \A	Returns a match if the specified characters are at the beginning of the string
# \A is case sensitive
x = re.findall("\AHi", txt)

print(x)

if x:
  print("Yes, there is a match!")
else:
  print("No match")
```

```python
#\b	Returns a match where the specified characters are at the beginning or at the end of a word
#(the "r" in the beginning is making sure that the string is being treated as a "raw string")

txt = "Hi all How are you doing, Doing well"

x = re.findall(r"\bll", txt)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")

x = re.findall(r"ll\b", txt)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```

```python
# \d - Check if the string contains any digits (numbers from 0-9):
# this gives every digit character

txt1 = "Now the time is 7:00pm"
x = re.findall("\d", txt1)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```

```python
# \D	Returns a match where the string DOES NOT contain digits

#Return a match at every no-digit character:

x = re.findall("\D", txt1)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```

```python
# [arn]	Returns a match where one of the specified characters (a, r, or n) is present

txt = "Hi all How are you doing, Doing well"

x = re.findall("[arn]", txt) # these are case sensitive patterns

print(x)
```

```python
# [a-n]	Returns a match for any lower case character, alphabetically between a and n
x = re.findall("[a-n]", txt)

print(x)
```

```python
# [^arn]	Returns a match for any character EXCEPT a, r, and n
x = re.findall("[^arn]", txt)

print(x)
```

```python
txt = "08 times before 11:45 AM"
# [0123]	Returns a match where any of the specified digits (0, 1, 2, or 3) are present
x = re.findall("[0123]", txt)

print(x)
```

```python

# [0-9]	Returns a match for any digit between 0 and 9
x = re.findall("[0-9]", txt) # these are single character/digits

print(x)
```

```python
# [0-5][0-9]	Returns a match for any two-digit numbers from 00 and 59
x = re.findall("[0-5][0-9]", txt) # return all 2 digits numbers where tens position is having 0-5, and ones pos is 0-9
# understand the difference between 1 and 01

print(x)
```

```python
# [a-zA-Z]	Returns a match for any character alphabetically between a and z, lower case OR upper case
txt = "Aw PQ rT oy"
x = re.findall("[A-Za-z]", txt)

print(x)
```

**Python JSON**

JavaScript object notation

JSON is a syntax for storing and exchanging data.

The de facto standard for information exchange.

JSON supports primitive types, like strings and numbers, as well as nested lists and objects.

Lets try to read, write JSON in Python.

```python
# Python JSON to dict
import json

person = '{"name": "Bob", "languages": ["English", "French"]}' # this is a string read from a file
print(type(person))
person_dict = json.loads(person) # loading str to dict(JSON format)

# Output: {'name': 'Bob', 'languages': ['English', 'French']}
print( person_dict)
print(type(person_dict))

# Output: ['English', 'French']
print(person_dict['languages'])
```

```python
import json

data = {
    "president": {
        "name": "Zaphod Beeblebrox",
        "species": "Betelgeusian"
    }
}
print(type(data))
print(data)

with open("data_file.json", "w") as write_file: # extension .json, with is for a file having newlines, everytime new line is considered
    json.dump(data, write_file)

json_string = json.dumps(data,indent=4) # indent needed to read in indent format

print(json_string)
print(type(json_string))
```

```python
# a Python object (dict):
x = {
  "name": "John",
  "age": 30,
  "city": "New York"
}

# convert into JSON:
y = json.dumps(x)

# the result is a JSON string:
print(y)
```

```python
print(json.dumps({"name": "John", "age": 30}))
print(json.dumps(["apple", "bananas"]))
print(json.dumps(("apple", "bananas")))
print(json.dumps("hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
```

```python
x = {
  "name": "John",
  "age": 30,
  "married": True,
  "divorced": False,
  "children": ("Ann","Billy"),
  "pets": None,
  "cars": [
    {"model": "BMW 230", "mpg": 27.5},
    {"model": "Ford Edge", "mpg": 24.1}
  ]
}

print(json.dumps(x,indent = 4))

print(json.dumps(x, indent=4, sort_keys = True))
```

**Pickling in Python**

Pickling - converts any kind of python objects (list, dict, etc.) into byte streams (0s and 1s) is called pickling or serialization or flattening or marshalling

Unpickling - converts the byte stream (generated through pickling) back into python objects

Why ?? - Easy to transfer data from One server to another and store it in a file

Beware of untrusted source

```python
import pickle

def storeData():
    # initializing data to be stored in db
    Omkar = {'key' : 'Omkar', 'name' : 'Omkar Pathak',
    'age' : 21, 'pay' : 40000}
    Jagdish = {'key' : 'Jagdish', 'name' : 'Jagdish Pathak',
    'age' : 50, 'pay' : 50000}

    # database
    db = {}
    db['p1'] = Omkar
    db['p2'] = Jagdish

    # Its important to use binary mode
    dbfile = open('example.pickle', 'ab') # ab : append in binary mode

    # source, destination
    pickle.dump(db, dbfile)
    dbfile.close()

def loadData():
    # for reading also binary mode is important
    dbfile = open('example.pickle', 'rb')  # rb : read n binary mode
    db = pickle.load(dbfile)
    for keys in db:
        print(keys, '=>', db[keys])
    dbfile.close()

if __name__ == '__main__': # this is the syntax
    print("abc")
    storeData()
    loadData()
```

```python

# initializing data to be stored in db
Omkar = {'key' : 'Omkar', 'name' : 'Omkar Pathak',
'age' : 21, 'pay' : 40000}
Jagdish = {'key' : 'Jagdish', 'name' : 'Jagdish Pathak',
'age' : 50, 'pay' : 50000}

# database
db = {}
db['Omkar'] = Omkar
db['Jagdish'] = Jagdish

# For storing
b = pickle.dumps(db)       # type(b) gives <class 'bytes'>
print(type(b))
print(b)

# For loading
myEntry = pickle.loads(b)
print(myEntry)
print(type(myEntry))
```
