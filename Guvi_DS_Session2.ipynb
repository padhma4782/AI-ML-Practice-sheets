```python
# Input : List of integers
# For each of the element in the list check if its a positive number and check if its a multiple of 2

nums = [2,4,-1]

#func1 : check for a number is Positive or not
def func1(x):
  if x>= 0:
    print("Positive")
  else:
    print("Negative")

# func2 : Check for a number is multiple of num or not
def func2(y,num):
  if y % num ==0:
    print("Multiple of ",num)
  else:
    print("Not multiple of ",num)

for x in nums:
  print(x)
  func1(x)
  func2(x,4)
```
