```python
# List Slicing
var1 = [1,5,3,78,90,54,34,45]

print(var1[2:4])
print(var1[2:5]) # slices the string from index 2 to index 5 (not including 5)
print(var1[:5]) # slices from starting till index5(not including 5)
print(var1[2:]) # Slices from 2 till end

print(var1[-5:]) # slice from -5 till end
print(var1[-5:-2]) # slice from -5 till -2 (not including -2)
print(var1[:-3]) # slice from starting till -3(not included)

print(var1[2:-1]) # mixing of indexing and -ve index is allowed in slicing
```
