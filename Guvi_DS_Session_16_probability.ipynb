```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random float numbers between 0 and 1 (inclusive of 0, exclusive of 1)
uniform_samples = np.random.uniform(2, 5, 100000)

# Create a histogram
plt.hist(uniform_samples, bins=20, alpha=0.75, rwidth=0.85)

# Set labels and title
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of 100000 Uniform Random Numbers in [0, 1)')

# Show the histogram
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the scale (lambda) parameter for the exponential distribution
scale_parameter = 100

# Generate 1000 random numbers from an exponential distribution
exponential_samples = np.random.exponential(scale_parameter, 100000)

# Create a histogram
plt.hist(exponential_samples, bins=20, alpha=0.75, rwidth=0.85)

# Set labels and title
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of 1000 Random Numbers from Exponential Distribution')

# Show the histogram
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the mean and standard deviation for the normal distribution
mean = 10
std_deviation = 5.0

# Generate 1000 random numbers from a normal distribution
normal_samples = np.random.normal(mean, std_deviation, 100000)

# Create a histogram
plt.hist(normal_samples, bins=20, alpha=0.75, rwidth=0.85)

# Set labels and title
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of 1000 Random Numbers from Normal Distribution')

# Show the histogram
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
population_size = 100000  # Size of the population
sample_size = 40  # Size of each sample
num_iter = 10000  # Number of samples to generate

# Generate samples from a uniform distribution (population)
#population = np.random.uniform(0, 1, population_size)
#population = np.random.exponential(1, population_size)
population = np.random.normal(3, 2, population_size)

# Initialize an array to store sample means
sample_means = []

# Generate samples and calculate sample means
for _ in range(num_iter):
    sample = np.random.choice(population, size=sample_size, replace=True)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)
    #print(sample,sample_mean,sample_means)

# Plot the distribution of sample means
plt.hist(sample_means, bins=30, density=True, alpha=0.6, color='b', label='Sample Means')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.title('Central Limit Theorem Simulation')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sample_chi_squared(degrees_of_freedom, sample_size):
    # Sample from chi-squared distribution
    samples = np.random.chisquare(degrees_of_freedom, size=sample_size)
    return samples

def plot_chi_squared_histogram(samples, degrees_of_freedom):
    # Plot histogram
    plt.hist(samples, bins='auto', alpha=0.7, color='blue', edgecolor='black')
    plt.title(f'Chi-Squared Distribution (Degrees of Freedom: {degrees_of_freedom})')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.show()

# Specify the degrees of freedom and sample size
degrees_of_freedom_user = int(input("Enter the degrees of freedom: "))
sample_size_user = int(input("Enter the sample size: "))

# Sample from chi-squared distribution
chi_squared_samples = sample_chi_squared(degrees_of_freedom_user, sample_size_user)

# Plot histogram
plot_chi_squared_histogram(chi_squared_samples, degrees_of_freedom_user)
```
