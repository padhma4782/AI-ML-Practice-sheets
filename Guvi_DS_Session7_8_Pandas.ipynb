**Pandas** :

Python library used for working with data sets.

It has functions for analyzing, cleaning, exploring, and manipulating data.

The name "Pandas" has a reference to both "Panel Data", and "Python Data Analysis" and was created by Wes McKinney in 2008.

Pandas allows us to analyze big data and make conclusions based on statistical theories.

Pandas can clean messy data sets, and make them readable and relevant.

Relevant data is very important in data science.

Pandas gives you answers about the data. Like:

Is there a correlation between two or more columns?

What is average value?

Max value?

Min value?

Pandas are also able to delete rows that are not relevant, or contains wrong values, like empty or NULL values. This is called cleaning the data.

```python
# Handliing Missing data

df = pd.read_csv("weather_data.csv",parse_dates=['day']) # make the day colums as date format, else it will be str
print(type(df.day[0]))
df
```

```python
df.set_index('day',inplace=True)
df
```

```python
new_df = df.fillna(0) # in the whole df change NaN to 0, its not an inplace operation
new_df
#df
```

```python
# Column specific nan
new_df = df.fillna({
        'temperature': 0,
        'windspeed': 0,
        'event': 'No Event'
    })
new_df
```

```python
# Carry forward the previous data
new_df = df.fillna(method="ffill") # forward fil
new_df
```

```python
# use the next data

new_df = df.fillna(method="bfill") # backward fill
new_df
```

```python
new_df = df.fillna(method="bfill", axis="columns") # axis is either "index" or "columns" , copies values horizontally
new_df
```

```python
new_df = df.fillna(method="ffill",limit=1) # limitting carry forward only once, multiple carry forward not allowed
new_df
```

```python
# intrapolate
new_df = df.interpolate() # linear intrapolation : Works for numbers
new_df
```

```python
new_df = df.interpolate(method="time") # there are some missing days from 01 to 04 so taking care of intrapolation near to 05
new_df
```

```python
# dropna

new_df = df.dropna() # atleast one na then drop
print(df)
new_df
```

```python
new_df = df.dropna(how='all') # drop only when all are na
new_df
```

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35, 38],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "humidity": [65,68,71],
})
df2
```

```python
df3=pd.merge(df1,df2,on="city",how="inner") # look at the common only : df1 intersection df2
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="outer") # df1 union df2
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="left") # merge wrt to left arg : df1
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="right") # merge wrt to right arg : df2
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="outer",indicator=True) # indicator of merge
df3
```

```python
import pandas as pd

# Creating a DataFrame with string dates
data = {'date': ['2021-01-01', '2021-01-02', '2021-01-03'], 'value': [100, 200, 300]}
df = pd.DataFrame(data)

# Converting 'date' column to datetime
df['date'] = pd.to_datetime(df['date'])

print(df)
```

```python
# Generating Range of Dates
# Generating daily data starting from January 1, 2021
date_range = pd.date_range(start='1/1/2021', periods=10, freq='M')
print(date_range)
```

```python
# Suppose df contains daily data, and you want to convert it to monthly data by taking the mean.
monthly_df = df.resample('M').mean()
print(monthly_df)
```

```python
# Set timezone
date_range_tz = pd.date_range(start='1/1/2021', periods=10, freq='D', tz='UTC')
print(date_range_tz)

# Convert timezone
date_range_tz = date_range_tz.tz_convert('America/New_York')
print(date_range_tz)

```

```python
from pandas.tseries.offsets import BDay

# Business day offset (adds one business day)
today = pd.Timestamp('2024-04-15')
next_business_day = today + BDay(1)
print(next_business_day)
```

```python
from pandas.tseries.holiday import USFederalHolidayCalendar

# US Federal Holiday Calendar
cal = USFederalHolidayCalendar()
holidays = cal.holidays(start='2021-01-01', end='2021-12-31')
print(holidays)
```
